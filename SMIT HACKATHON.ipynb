```python
import json
import os
from datetime import date
import pandas as pd

class HCMLiteNotebookSimulator:
    def __init__(self, storage_file="hcm_storage.json"):
        self.storage_file = storage_file
        self.init_storage()

    def init_storage(self):
        if not os.path.exists(self.storage_file):
            self.db = {
                "hcm_company": {},
                "hcm_users": [],
                "hcm_attendance": [],
                "hcm_leaves": []
            }
            self.save_data()
        else:
            with open(self.storage_file, "r") as f:
                self.db = json.load(f)

    def save_data(self):
        with open(self.storage_file, "w") as f:
            json.dump(self.db, f, indent=4)

    def setup_company(self, company_name):
        self.db["hcm_company"] = {
            "company_name": company_name,
            "office_start_time": "09:00",
            "grace_minutes": 15,
            "annual_quota": 14
        }
        self.save_data()

    def add_user(self, name, email, role, department, salary):
        user = {
            "id": len(self.db["hcm_users"]) + 1,
            "name": name,
            "email": email,
            "role": role,  # admin, accountant, employee
            "department": department,
            "salary": float(salary)
        }
        self.db["hcm_users"].append(user)
        self.save_data()

    def record_attendance(self, employee_id, check_in_time, status):
        att = {
            "employee_id": employee_id,
            "date": str(date.today()),
            "check_in": check_in_time,
            "status": status  # present, late, absent
        }
        self.db["hcm_attendance"].append(att)
        self.save_data()

    def record_leave(self, employee_id, leave_type, days_count, status):
        leave = {
            "employee_id": employee_id,
            "leave_type": leave_type,  # casual, sick
            "days": days_count,
            "status": status  # approved, pending
        }
        self.db["hcm_leaves"].append(leave)
        self.save_data()

# POPULATING SAMPLE DATA
sim = HCMLiteNotebookSimulator()
sim.setup_company("Saylani IT Park")

# Adding Admin, Accountant, and Employees
sim.add_user("Ali Admin", "admin@saylani.com", "admin", "Management", 0)
sim.add_user("Sara Accountant", "accountant@saylani.com", "accountant", "Finance", 75000)
sim.add_user("Ahmed Khan", "ahmed@saylani.com", "employee", "Web Development", 50000)
sim.add_user("Fatima Ali", "fatima@saylani.com", "employee", "Web Development", 55000)
sim.add_user("Usman Raza", "usman@saylani.com", "employee", "Accounts", 48000)

# Simulating attendance & leaves
sim.record_attendance(3, "08:55", "present")
sim.record_attendance(4, "09:20", "late")
sim.record_attendance(5, "09:05", "present")
sim.record_leave(3, "casual", 2, "approved")
sim.record_leave(4, "sick", 1, "approved")


# TEXT & TABLE-BASED ADMIN DASHBOARD
users_df = pd.DataFrame(sim.db["hcm_users"])
attendance_df = pd.DataFrame(sim.db["hcm_attendance"])
leaves_df = pd.DataFrame(sim.db["hcm_leaves"])

print("         SAYLANI HCM LITE - ADMIN DASHBOARD       ")

total_employees = len(users_df[users_df['role'] == 'employee'])
total_present = len(attendance_df[attendance_df['status'] == 'present'])
total_late = len(attendance_df[attendance_df['status'] == 'late'])
total_payroll = users_df['salary'].sum()

print(f" Total Active Employees : {total_employees}")
print(f"Present Today          : {total_present}")
print(f" Late Today             : {total_late}")
print(f"Total Monthly Payroll  : PKR {total_payroll:,.2f}")
print("--------------------------------------------------")

print("\n 1. EMPLOYEE DIRECTORY ")
display(users_df[['id', 'name', 'role', 'department', 'salary']])

print("\n 2. TODAY'S ATTENDANCE LOGS ")
if not attendance_df.empty:
    display(attendance_df)

print("\n 3. LEAVE RECORDS ")
if not leaves_df.empty:
    display(leaves_df)
```
